```python
counts = df.Rating.value_counts()
```

```python
counts_df = pd.DataFrame ({'calif': counts.index, 'frq': counts.values})
counts_df.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.barplot(
    data = counts_df, 
    x= 'calif',
    y= 'frq'
)
plt.title('Frecuencia_de_calificaciones')
plt.show()
          
```
